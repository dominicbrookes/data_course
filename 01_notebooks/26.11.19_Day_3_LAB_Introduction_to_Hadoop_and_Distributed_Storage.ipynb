## Running Ironhack notebooks in VS Code

### One-off setup (already done)

Your `.venv_airflow` environment already has `ipykernel` and `pandas` installed, so you don't need to repeat this. It's here in case you ever make a new environment or need a new package.

1. In the VS Code terminal, make sure you're in `~/data_course`.
2. Switch the environment on: `source .venv_airflow/bin/activate` (the prompt should start with `(.venv_airflow)`).
3. Install what notebooks need: `pip install ipykernel pandas`
4. For any other package a lab needs later (for example numpy), use the same pattern: `pip install <package name>`

### Every time you add a new notebook

Each new .ipynb file needs its kernel picking once. After that, VS Code remembers it for that file.

1. **Drag it in.** Drag the downloaded .ipynb from Windows into your `01_notebooks` folder in the VS Code Explorer (left-hand panel). Rename it if you like (right-click → Rename), for example replacing DATE with the date.
2. **Open it.** Click the file in the Explorer.
3. **Click Select Kernel.** It's in the top-right corner of the notebook. (If it already says `.venv_airflow`, skip to step 7.)
4. **Click Python Environments…**
5. **Pick .venv_airflow.** Choose the line ending `.venv_airflow/bin/python`. Ignore the one marked "Recommended" (`/usr/bin/python3`): that's Ubuntu's own Python and it won't work.
6. **Check the top-right.** It should now say `.venv_airflow`.
7. **Run the import cell.** Click ▷ next to the import cell. A green tick and [1] mean it's working.

### Q. 1 — Scale Up or Scale Out?

Complete the table.

| Change | Scale up or scale out? |
|---|---|
| Add more RAM to one server | |
| Add four worker machines | |
| Replace one CPU with a faster CPU | |
| Add three storage nodes | |

========================================================================
ANSWER Q1
========================================================================

'So making a change to one machine (server) is up and adding in more machines (servers) is out. In reality, you rent the space from a cloud service provider like Amazon (AWS) so you're renting a small piece of a server (a virtual machine) and if you scale 'up', you're asking for a bigger share of processor or RAM from that machine. When you fill up your servers, you scale 'out' by adding more servers. If we swap CPU, we swap from one machine to a machine with a faster CPU etc. (or, on a physical server, fit a faster CPU into the same machine)...but we go from one to one, which is why this is scale up. Adding more virtual machines can still be staying on one physical machine, but renting more of that machine, renting more virtual servers, which is different from asking for a bigger share for your existing virtual machine. Each new virtual machine counts as another server in the cluster, that's why adding storage nodes, adding more virtual machines is scaling out.'

========================================================================
========================================================================

Change: Add more RAM to one server

Scale up or scale out?: Scale up. 

====================================

Change: Add four worker machines

Scale up or scale out?: Scale out.

====================================

Change: Replace one CPU with a faster CPU

Scale up or scale out?: Scale up.

====================================

Change: Add three storage nodes

Scale up or scale out?: Scale out.

### Q. 3 — Calculate Blocks with Python

Use:

```python
File size = 1,025 MB
Block size = 128 MB
```

Write Python code to calculate the number of blocks.

````python
file_size_input = int(input("The File size is :"))
block_size_input = int(input("The Block size is :"))

def calculator():
    full_answer = (file_size_input // block_size_input)
    if (file_size_input % block_size_input) > 0:
        full_answer = full_answer + 1
    else: full_answer
    return(full_answer)
print(f"The answer to the question is: {calculator()}.")
# The brackets after a function name list what it needs to be given (parameters). Empty brackets, as in def calculator(), mean it needs nothing passed in, 
# because it reads file_size_input and block_size_input from the top of the cell. The brackets in def and in the call must match: def calculator(question) needs one thing 
# passed in, like calculator(1), even if it's never used. You always need the brackets when calling, even empty, because calculator() runs the function and calculator on 
# its own doesn't.
````

### Q. 17 — Compare Two Replication Choices

A 5 TB dataset can use:

```python
replication factor 2
replication factor 3
```

Calculate approximate stored data for both choices and explain the trade-off.

````python
file_size = 5
replication_factor_2 = 2
replication_factor_3 = 3

q_seventeen_a = file_size * replication_factor_2
q_seventeen_b = file_size * replication_factor_3

print(f"The approxiamte stored data for replication factor 2 is {q_seventeen_a} TB.")
print(f"The approxiamte stored data for replication factor 3 is {q_seventeen_b} TB.")

# Replication factor 3 uses more storage but provides another copy of each block. A block is a piece of one file, not a group of files.
# Your code never gets split into blocks. Only the data files you store in HDFS are split, like a 700 MB sales CSV.
# A full copy of your code is sent to every server that holds a block, and each server runs it on its own block.
# Example: total sales per city across a file split into 3 blocks.
# Map: each server reads its own block and works out subtotals per city, e.g. "Leeds £500, Manchester £300".
# Shuffle: all the Leeds subtotals are sent to one server, all the Manchester subtotals to another, and so on.
# Reduce: the Leeds server adds up all the Leeds subtotals to get the final Leeds total. Same for every other city.
# Each row is read whole by exactly one task, so every sale is counted once and the final totals are accurate.
````
